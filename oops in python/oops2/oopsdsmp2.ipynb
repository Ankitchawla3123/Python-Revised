```python
# pass by reference 
class Person:
    def __init__(self,name,gender):
        self.name=name
        self.gender=gender

def greet(person):
    print("Hi my name is", person.name, "and i am a", person.gender)
    person.name="yo"
    # reference is passed on so even if we update in here it will change globally 

p=Person("ankit", "male")
greet(p)
p.name
```

```python
#Multablitiy of an OBJECT
class Person:
    def __init__(self,name,gender):
        self.name=name
        self.gender=gender

def greet(person):
    print("Hi my name is", person.name, "and i am a", person.gender)
    person.name="yo"
    # reference is passed on so even if we update in here it will change globally 
    return person

p=Person("ankit", "male")
p2=greet(p)
print(id(p))
print(id(p2))
# ID IS SAME FOR BOTH SO OBJECT IS MUTABLE i.e list set and dict are formed via this process
```
